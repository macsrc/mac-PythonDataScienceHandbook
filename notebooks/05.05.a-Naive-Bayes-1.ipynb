### 1. Spam Detection (Text Classification)
- **Classify** emails as spam or not spam.

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
```

```python
# Sample dataset
emails = ["Win money now", "Hello friend", "Claim your prize", "How are you?", "Free entry in contest"]
labels = [1, 0, 1, 0, 1]  # 1 = Spam, 0 = Not Spam
```

```python
# Convert text to numeric features
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(emails)
```

```python
# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(X, labels, test_size=0.3, random_state=42)
```

```python
# Train Naive Bayes model
nb = MultinomialNB()
nb.fit(X_train, y_train)
```

```python
# Predictions
y_pred = nb.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
```

### 2. Iris Flower Classification
- Classify iris flowers into species.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score
```

```python
# Load dataset
iris = load_iris()
X, y = iris.data, iris.target
```

```python
# X
# y
```

```python
# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
# X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=42)
```

```python
# Train Naive Bayes
model = GaussianNB()
model.fit(X_train, y_train)
```

```python
# Predict
y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
```

### 3. Titanic Survival Prediction (Categorical + Numeric)
- Use Naive Bayes for survival classification.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score
```

```python
# Load Titanic dataset (from seaborn or CSV)
import seaborn as sns
titanic = sns.load_dataset("titanic").dropna(subset=['age', 'sex', 'class', 'survived'])
```

```python
# titanic.info()
titanic.describe()
```

```python
# Preprocess
titanic['sex'] = titanic['sex'].map({'male': 0, 'female': 1})
titanic['class'] = titanic['class'].map({'First': 1, 'Second': 2, 'Third': 3})

X = titanic[['age', 'sex', 'class']]
y = titanic['survived']
```

```python
X
# y
```

```python
# Train-Test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Train NB
nb = GaussianNB()
nb.fit(X_train, y_train)
```

```python
# Predict
y_pred = nb.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
```
