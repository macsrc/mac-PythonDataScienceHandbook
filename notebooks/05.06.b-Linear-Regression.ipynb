```python
#1
%matplotlib inline
import matplotlib.pyplot as plt
plt.style.use('seaborn-v0_8-whitegrid')
import numpy as np
```

```python
#2
rng = np.random.RandomState(1)
x = 10 * rng.rand(50)
y = 2*x - 5 + rng.randn(50)
plt.scatter(x,y)
plt.show()
```

```python
## 3 ##
# 1. import LinearRegression
# 2. create model using LR and intercept values
# 3. fit using the model - x - np.newaxis && y 
# 4. define xfit - get linespace as (0, 10, 1000)
#     and yfit - model.predict with xfit[:, np.newaxis] 
# 5. plot it:
#     plt.scatter(x,y)
#     plt.plot(xfit, yfit)
from sklearn.linear_model import LinearRegression
model = LinearRegression(fit_intercept=True)

model.fit(x[:, np.newaxis], y)

xfit = np.linspace(0, 10, 1000)
yfit = model.predict(xfit[:, np.newaxis]) 

plt.scatter(x, y)
plt.plot(xfit, yfit)
plt.show()
```

### @04-Aug-2025

```python
## 4
print("Model slope_     :", model.coef_[0])
print("model intercept_ :", model.intercept_)
```

```python
## 5
rng = np.random.RandomState(1)
X = 10 * rng.rand(100, 3)
y = 0.5 + np.dot(X, [1.5, -2., 1.])

model.fit(X, y)
print(model.intercept_)
print(model.coef_)
```

```python
## 6 - polynamial basis function
from sklearn.preprocessing import PolynomialFeatures
x = np.array([2,3,4])
poly = PolynomialFeatures(3, include_bias=False)
poly.fit_transform(x[:, None])
```

```python
## 7 - pipeline
from sklearn.pipeline import make_pipeline
poly_model = make_pipeline(PolynomialFeatures(7), LinearRegression())
```

```python
## 8 
rng = np.random.RandomState(1)
x = 10 * rng.randn(50)
y = np.sin(x) + 0.1 * rng.randn(50)

poly_model.fit(x[:, np.newaxis], y)
yfit = poly_model.predict(xfit[:, np.newaxis])

plt.scatter(x, y)
plt.plot(xfit, yfit)
```
