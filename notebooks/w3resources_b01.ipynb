```python
# 55. Find Local IPs
# Write a Python program to find local IP addresses using Python's stdlib.
#### Solution Approach:
    # The following code retrieves the local IP address of the current machine:
    # 1. Use 'socket.gethostname()' to get the local hostname.
    # 2. Use 'socket.gethostbyname_ex()' to get a list of IP addresses associated with the hostname.
    # 3. Filter the list to exclude any IP addresses starting with "127." (loopback addresses).
    # 4. Extract the first IP address (if available) from the filtered list.
    # 5. Print the obtained IP address to the console.

import socket

local_hostname = socket.gethostname()
print(1, local_hostname)

ip_addresses = socket.gethostbyname_ex(local_hostname)[2]
print(ip_addresses)

filtered_ips = [ip for ip in ip_addresses if not ip.startswith("127.")]
print(filtered_ips)

first_ip = filtered_ips[:1]
print(first_ip)

print(first_ip[0])
```

```python
# # 65. Seconds to DHMS Converter
# Write a Python program that converts seconds into days, hours, minutes, and seconds.

# Prompt the user to input a time duration in seconds and convert it to a float.
time = float(input("Input time in seconds: "))

# Calculate the number of full days in the given time duration.
day = time // (24 * 3600)
# Update the time variable to hold the remaining seconds after subtracting full days.
time = time % (24 * 3600)

# Calculate the number of full hours in the remaining time.
hour = time // 3600
# Update the time variable to hold the remaining seconds after subtracting full hours.
time %= 3600

# Calculate the number of full minutes in the remaining time.
minutes = time // 60
# Update the time variable to hold the remaining seconds after subtracting full minutes.
time %= 60

# The 'time' variable now represents the remaining seconds, which is the number of seconds.
seconds = time

# Print the time duration in the format "d:h:m:s".
print("d:h:m:s-> %d:%d:%d:%d" % (day, hour, minutes, seconds))
```

```python
# 70. Sort Files by Date
# Write a Python program to sort files by date.
import os
from datetime import datetime

# Set the folder path
folder_path = "./"  # Change this to your folder

# Get list of files with full path
files = [os.path.join(folder_path, f) for f in os.listdir(folder_path) if os.path.isfile(os.path.join(folder_path, f))]

# Sort files by modification time
files.sort(key=lambda x: os.path.getmtime(x))

# Print sorted files with date
for f in files:
    mod_time = os.path.getmtime(f)
    print(f"{datetime.fromtimestamp(mod_time)} - {os.path.basename(f)}")
```

```python
# Here's a step-by-step breakdown of the code to **sort files by date** in a folder:
### ✅ **1. Import Required Modules**

import os
from datetime import datetime


# * `os`: For file and folder operations.
# * `datetime`: To format timestamps into readable dates.
```

```python
### ✅ **2. Set the Target Folder Path**

folder_path = "./"  # or specify like "C:/Users/YourName/Documents"

# * `./` means the current folder.
# * You can replace it with any full folder path.
```

```python
### ✅ **3. List All Files in That Folder**

files = [
    os.path.join(folder_path, f)
    for f in os.listdir(folder_path)
    if os.path.isfile(os.path.join(folder_path, f))
]

# * `os.listdir(folder_path)` → lists all files **and folders**.
# * `os.path.isfile()` → ensures we pick **only files**, not folders.
# * `os.path.join()` → joins folder path + file name to get full path.
print(files)
```

```python
### ✅ **4. Sort the Files by Modification Time**

files.sort(key=lambda x: os.path.getmtime(x))

# * `os.path.getmtime(x)` → gets the **last modified time** of file `x`.
# * `sort()` → sorts the list in **ascending order** (oldest to newest).
print(files)
```

```python
### ✅ **5. Print Sorted File Names and Dates**

for f in files:
    mod_time = os.path.getmtime(f)
    print(f"{datetime.fromtimestamp(mod_time)} - {os.path.basename(f)}")


# * `datetime.fromtimestamp(mod_time)` → converts timestamp to readable date.
# * `os.path.basename(f)` → gets only the file name (not full path).

### ✅ **Output Example**

# 2024-05-01 09:12:33 - report1.txt
# 2024-06-15 17:44:10 - image.png
# 2024-07-29 12:01:25 - notes.docx
```

```python
import shutil
import os

def copy_self():
    """
    Creates a copy of the current Python script's source code.
    The copy will be named 'copied_script.py' in the same directory.
    """
    try:
        # Get the path of the current script
        source_path = os.path.abspath(__file__)

        # Define the destination path for the copy
        # You can change 'copied_script.py' to any desired name
        destination_path = os.path.join(os.path.dirname(source_path), 'copied_script.py')

        # Copy the source file to the destination
        shutil.copyfile(source_path, destination_path)

        print(f"Source code copied successfully to: {destination_path}")

    except Exception as e:
        print(f"An error occurred: {e}")

if __name__ == "__main__":
    copy_self()
```
